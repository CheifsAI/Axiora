```python
import pandas as pd
from LLM import llama3b
from Visualizer import Visualizer
from DataAnalyzer import DataAnalyzer
```

```python
df = pd.read_csv("Test_Datasets/laptop_price.csv", encoding='latin1')
df.head(5)
```

```python
analyzer = DataAnalyzer(dataframe=df,llm=llama3b)
visualizer = Visualizer(dataframe=df)
```

```python
analysis = analyzer.analysis_data()
analysis
```

```python
print(analyzer.memory[1])
```

```python
questions = analyzer.questions_gen(4)
```

```python
ques_list = ['What is the average price of laptops sold by each company?',
              'How does the weight of laptops change with their screen resolution?',
                'What is the distribution of laptop types (TypeNames) and how do they relate to price and inches?', 
                'What is the average price-to-weight ratio for laptops from each company?']
```

```python
charts =[]
columns = []
for question in ques_list:
    charts.append(analyzer.select_chart_type(question))
    columns.append(analyzer.select_columns(question))
```
