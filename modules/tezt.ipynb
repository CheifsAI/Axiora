```python
df = pd.read_csv("Test_Datasets/laptop_price.csv", encoding='latin1')
df.head(5)
```

```python
analysis = analyzer.analysis_data()
analysis
```

```python
questions = analyzer.questions_gen(4)
```

```python
ques_list =['What is the average price of laptops and how does it compare to the median price?',
             'Which laptop models have an average screen resolution above 1920x1080 and their corresponding weight ranges?',
            'How does the operating system (OS) influence the price of laptops?', 
            'What is the correlation between the weight of laptops and their screen resolution?']
```

```python
charts =[]
columns = []
for question in ques_list:
    charts.append(analyzer.select_chart_type(question))
    columns.append(analyzer.select_columns(question))
```
