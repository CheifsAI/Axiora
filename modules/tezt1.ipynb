```python
ques_list = ['What is the average price of products sold?', 
             'How does revenue correlate with stock levels?',
              'What is the average shipping time for products?', 
              'How does manufacturing lead time affect production costs?']
```

```python
charts =[]
columns = []
for question in ques_list:
    charts.append(analyzer.select_chart_type(question))
    columns.append(analyzer.select_columns(question))
```
